```python
class Rectangle:
    def __init__(self, width, height):
        # Attributes
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)
```

```python
class Rectangle:
    def __init__(self, width, height):
        # Attributes
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    # Special methods
    def __str__(self):
        return f"Rectangle: width={self.width}, height={self.height})"

    def __repr__(self):
        return f"Rectangle({self.width},{self.height})"
        
```
